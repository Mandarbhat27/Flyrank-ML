## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

# Target or Proxy

The dataset does not contain a direct label indicating whether a page should be refreshed. Therefore, a proxy target is created.

The proxy uses observable signals such as CTR and average search position. Pages with low CTR and poor search position are likely to have higher refresh priority.

The output is a refresh priority score rather than a fixed yes/no prediction.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["refresh_priority_proxy"] = (
    df["avg_position"].fillna(100) - df["ctr"]
)

df[["avg_position", "ctr", "refresh_priority_proxy"]].head()
```

# Success Metric

The primary evaluation metric is Precision@K.

Precision@K measures how many of the top-ranked pages are genuinely good candidates for refreshing.

A useful model should place the highest-priority pages near the top of the ranking so that editors spend time on the most valuable content.

# Why ML Beats a Fixed Rule

A fixed rule such as refreshing every page with CTR below 1% ignores many other useful signals.

Machine learning can combine several features, including impressions, average position, content age, engagement rate, and update history, to estimate refresh priority more accurately.

This produces better recommendations than relying on a single manually chosen threshold.
